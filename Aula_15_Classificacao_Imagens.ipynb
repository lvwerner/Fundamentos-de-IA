# 01. CLASSIFICAÇÃO DE IMAGEM

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread('/content/content/images.jpg')


# Load the cascade classifier using the full path
detec = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
cinza = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Detect faces
face = detec.detectMultiScale(cinza, 1.3, 3)
for (x, y, larg, alt) in face:
    ret = cv2.rectangle(img, (x, y), (x + larg, y + alt), (0, 255, 0), 3)
cv2_imshow(img)
cv2.waitKey(0)
cv2.destroyAllWindows() #
```

```python
import cv2
from google.colab.patches import cv2_imshow

# 1. Carregar a imagem
imagem = cv2.imread('/content/content/images.jpg')
imagem_gray = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

# 2. Carregar o classificador Haar Cascade para rosto
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# 3. Detectar rostos
rostos = face_cascade.detectMultiScale(imagem_gray, scaleFactor=1.1, minNeighbors=5)

# 4. Desenhar retângulos nos rostos detectados
for (x, y, w, h) in rostos:
    cv2.rectangle(imagem, (x, y), (x+w, y+h), (255, 0, 0), 2)
    cv2.putText(imagem, "P", (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 255), 2)

# 5. Saída da imagem
cv2_imshow(imagem)
cv2.waitKey(0)
cv2.destroyAllWindows()
```


# Manipulação de Imagens e Criação de Gráficos


- Conversão de espaços de cores
- Criação de histogramas
- Transformações geométricas
- Aplicação de filtros
- Detecção de bordas


```python
import matplotlib.pyplot as plt

plt.hist(gray.ravel(), bins=256)
plt.title("Histograma da Imagem")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de Pixels")
plt.show()
```

```python
resize = cv2.resize(img, (300,300))

plt.imshow(cv2.cvtColor(resize, cv2.COLOR_BGR2RGB))
plt.title("Imagem Redimensionada")
plt.axis("off")
plt.show()
```

```python
(h, w) = img.shape[:2]

M = cv2.getRotationMatrix2D((w/2, h/2), 45, 1.0)
rotacionada = cv2.warpAffine(img, M, (w, h))

plt.imshow(cv2.cvtColor(rotacionada, cv2.COLOR_BGR2RGB))
plt.title("Imagem Rotacionada")
plt.axis("off")
plt.show()
```

```python
blur = cv2.GaussianBlur(img, (15,15), 0)

plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.title("Imagem com Gaussian Blur")
plt.axis("off")
plt.show()
```

```python
bordas = cv2.Canny(gray, 100, 200)

plt.imshow(bordas, cmap='gray')
plt.title("Detecção de Bordas")
plt.axis("off")
plt.show()
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

fig, ax = plt.subplots(1,4, figsize=(18,5))

ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
ax[0].set_title("RGB")
ax[0].axis("off")

ax[1].imshow(gray, cmap='gray')
ax[1].set_title("Gray")
ax[1].axis("off")

ax[2].imshow(hsv)
ax[2].set_title("HSV")
ax[2].axis("off")

ax[3].imshow(lab)
ax[3].set_title("LAB")
ax[3].axis("off")

plt.show()
```
